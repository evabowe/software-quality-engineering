## Judge-Modus wählen



```python
JUDGE_MODE = "objective"  # "strict" | "objective" | "lenient"
```

```python
from pathlib import Path

ROLE_PROMPTS_DIR = Path("role_prompts")
JUDGE_MODES = ("strict", "objective", "lenient")

def load_role_prompt(mode: str) -> str:
    """Read the role prompt for the given judge mode from role_prompts/."""
    if mode not in JUDGE_MODES:
        raise ValueError(f"Unknown judge mode {mode!r}, expected one of {JUDGE_MODES}")
    return (ROLE_PROMPTS_DIR / f"{mode}_prompt.md").read_text(encoding="utf-8")

role_prompt = load_role_prompt(JUDGE_MODE)
print(f"Loaded {JUDGE_MODE!r} role prompt ({len(role_prompt)} characters)")
```

## Code-Beispiele vergleichen: Komplexität (complexipy)

Für jedes Code-Beispiel in `code_with_defects/` wird die **kognitive Komplexität** gemessen: ein Wert dafür, wie schwer Code für Menschen zu verstehen ist.

https://complexipy.com/understanding-scores/

```python
%pip install -q complexipy

from complexipy import file_complexity

CODE_DIR = Path("code_with_defects")

# Alle Code-Beispiele (ohne Tests und __init__.py)
code_files = sorted(p for p in CODE_DIR.glob("*.py") if p.name != "__init__.py")

print(f"{'Beispiel':<20} {'Funktionen':>10} {'Komplexität':>12} {'max. pro Funktion':>18}")
for code_file in code_files:
    functions = file_complexity(str(code_file)).functions
    total = sum(f.complexity for f in functions)
    highest = max((f.complexity for f in functions), default=0)
    print(f"{code_file.stem:<20} {len(functions):>10} {total:>12} {highest:>18}")
```
