## Code-Beispiele vergleichen: Komplexität (complexipy)

Für jedes Snippet in `snippets/` wird die **kognitive Komplexität** gemessen: ein Wert dafür, wie schwer Code für Menschen zu verstehen ist.
Die Stufen L1–L5 sind manuell gesetzt (siehe `meta.yaml`); `bank_account` hat keine Stufe, weil es als Kontrolle für Data Contamination dient.

https://complexipy.com/understanding-scores/

Voraussetzung: im Projektordner einmal `pip install -e .` ausführen. Die Werte in `meta.yaml` schreibt `python tools/compute_complexity.py`.

```python
import pandas as pd
from complexipy import file_complexity

from judge_lab.config import SNIPPETS_DIR
from judge_lab.snippets import load_snippets

rows = []
for snippet in load_snippets():
    functions = file_complexity(str(SNIPPETS_DIR / snippet.id / snippet.code_file)).functions
    rows.append({
        "Snippet": snippet.id,
        "Stufe": snippet.level or "–",
        "Funktionen": len(functions),
        "Kognitive Komplexität": sum(f.complexity for f in functions),
        "max. pro Funktion": max((f.complexity for f in functions), default=0),
        "LOC": snippet.loc,
        "in meta.yaml": snippet.cognitive,
    })

pd.DataFrame(rows).set_index("Snippet")
```
